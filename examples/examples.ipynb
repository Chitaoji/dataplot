```python
n = 10000

dp.random_seed(42)

x1 = dp.randn(n)
x2 = dp.randn(n)
x3 = dp.randn(n)
x4 = dp.randn(n)
x5 = dp.randn(n)
```

```python
hist = x1.join(x1 + x2).sample(2000, "random").hist(alpha=0.8)
hist
```

```python
import pandas as pd

line = x1.cumsum().plot(
    pd.date_range("20260101", "20270221 15", freq="1h"),
    reference_lines=["x=20260801", "x=20261215"],
    rolling=[1, 1000, 10000],
)
line
```

```python
scatter = (x1 + x2).sample(1000).scatter(x1.sample(1000), fit=True, alpha=0.5)
scatter
```

```python
hexbin = (x1 + x2).hexbin(x1)
hexbin
```

```python
qqplot = x1.join(x1.exp()).qqplot()
qqplot
```

```python
ksplot = (x1 + x2 + x3).ksplot()
ksplot
```

```python
ppplot = x1.join(x1 + x2 + x3).ppplot()
ppplot
```

```python
heatmap = x1.join(
    x1 + x2, x1 + x2 + x3, x1 + x2 + x3 + x4, x1 + x2 + x3 + x4 + x5
).corrmap()
heatmap
```

```python
dp.figure(
    hist,
    line,
    scatter,
    hexbin,
    qqplot,
    ksplot,
    ppplot,
    heatmap,
    title="Figure Collections",
    subplots_adjust={"top": 0.92},
)
```
